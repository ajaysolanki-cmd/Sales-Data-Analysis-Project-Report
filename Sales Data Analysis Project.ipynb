```python
df1['ORDERDATE'] = pd.to_datetime(df1['ORDERDATE'])
```

```python
df1['YEAR'] = df1['ORDERDATE'].dt.year
df1['MONTH'] = df1['ORDERDATE'].dt.month
df1['DAY'] = df1['ORDERDATE'].dt.day
```

```python
df1['SALES_PER_ITEM'] = df1['SALES'] / df1['QUANTITYORDERED']
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df1['SALES'], kde=True)
plt.title("Sales Distribution")
plt.show()
```

```python
sns.histplot(df1['QUANTITYORDERED'], kde=True)
plt.show()
```

```python
sns.countplot(x='DEALSIZE', data=df1)
plt.show()
```

```python
plt.figure(figsize=(12,5))
sns.countplot(x='PRODUCTLINE', data=df1)
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(12,5))

sns.barplot(
    x='PRODUCTLINE',
    y='SALES',
    data=df1
)

plt.xticks(rotation=45)
plt.show()
```

```python
country_sales = df1.groupby('COUNTRY')['SALES'].sum().sort_values(ascending=False)

country_sales.plot(kind='bar', figsize=(12,5))
plt.show()
```

```python
sns.boxplot(x='DEALSIZE', y='SALES', data=df1)
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(
    x='STATUS',
    y='SALES',
    data=df1
)

plt.xticks(rotation=45)
plt.show()
```

```python
monthly_sales = df1.groupby('MONTH')['SALES'].sum()

monthly_sales.plot(
    kind='line',
    marker='o',
    figsize=(10,5)
)

plt.title("Monthly Sales Trend")
plt.ylabel("Sales")
plt.show()
```

```python
yearly_sales = df1.groupby('YEAR')['SALES'].sum()

yearly_sales.plot(kind='bar')
plt.show()
```

```python
quarter_sales = df1.groupby('QTR_ID')['SALES'].sum()

quarter_sales.plot(kind='bar')
plt.show()
```

```python
top_customers = df1.groupby('CUSTOMERNAME')['SALES'].sum().sort_values(ascending=False).head(10)

top_customers.plot(kind='bar', figsize=(12,5))
plt.title("Top 10 Customers")
plt.show()
```

```python
product_sales = df1.groupby('PRODUCTLINE')['SALES'].sum().sort_values(ascending=False)

product_sales.plot(kind='bar', figsize=(12,5))
plt.show()
```

```python
top_products = df1.groupby('PRODUCTCODE')['QUANTITYORDERED'].sum().sort_values(ascending=False).head(10)

top_products.plot(kind='bar')
plt.show()
```
